```python
import pandas as pd

from inspections import features
from inspections.load import load_rds

pub = features.build_city_table("published")
pub.attrs["join_counts"]
```

```python
# Monthly volume and critical-violation rate: training ends 2014-04-16, test starts 2014-09-02
monthly = pub.groupby(pub["Inspection_Date"].dt.to_period("M"))["criticalFound"].agg(["size", "mean"])
monthly.plot(subplots=True, figsize=(9, 5));
```

```python
# Critical rate by inspector cluster in training data (the 14-fold spread studied in section 3)
train = pub[~pub["Test"]]
train.groupby("sanitarian")["criticalFound"].agg(["size", "mean"]).sort_values("mean")
```
